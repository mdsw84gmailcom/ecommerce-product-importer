## Findings

- The price list contains 27,125 article variants.
- The dataset contains 600 unique product models.
- The price list and master list both contain 600 unique models.
- A direct model comparison is required to confirm that both datasets contain the same model identifiers.
- Pricing information is complete for all article variants.
- Packaging-related fields are only populated for subset of products.
- Missing values in the `Färg` column are not always true missing values.
- Several AW26 products store color information in `Beskrivning 2`.
